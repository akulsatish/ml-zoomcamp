```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
df.head()
```

```python
#Q2 - How many records are in the dataset?

df.count() #Q.2 = 10,000. There are missing values, but most of them say 10000 so I suspect it is 10000

df.shape #Q.2 = 10,000. Again this shows that there are 10000 rows/entries/records
```

```python
#Q3 - How many fuel types are presented in the dataset?

df['fuel_type'].unique() #There are 3 types

#Answer = 3
```

```python
#Q4 - How many columns have missing values?

df.isnull().sum() #From the course

sum(df.isnull().any()) #Any helps with checking if there are ANY missing values. Summing them gives the direct answer

#Answer = 2
```

```python
#Q5 - What's the maximum fuel effeciency of cars from Asia?

df[df['origin'] == 'Asia'].fuel_efficiency_mpg.max()

#Answer = 41.2
```

```python
#Q6 - Median of horsepower

#6.1 - Find the median value of the horsepower column in the dataset

print(f"Median HorsePower is: {df['horsepower'].median()}")

#6.2 - Find the most frequent value
hp_mode = df['horsepower'].mode()[0]
print(f"Mode is: {hp_mode}")

#6.3 - Fill missing values with mode
df['horsepower_filled'] = df['horsepower'].fillna(hp_mode)

#6.4 - Calculate median after dealing with missing values
print(f"Median horsepower after filling missing is: {df['horsepower_filled'].median()}")
```

```python
#Q7 - Sum of Weights

#7.1 - Select all the cars from Asia

asia_df = df[df['origin'] == 'Asia']

#7.2 - Select only columns vehicle_weight and model_year

asia_df_vehicle_model_df = asia_df[['vehicle_weight', 'model_year']]

#7.3 + 7.4 - Select only the first 7 values and store in np array

X = asia_df_vehicle_model_df.iloc[:7].to_numpy()

#7.4 - Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
xtx = np.dot(X.T, X)

#7.5 - Invert xtx
xtx_inv = np.linalg.inv(xtx)

#7.6 - Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = [1100, 1300, 800, 900, 1000, 1100, 1200]

#7.7 - Mutlitply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w
w = np.dot(np.dot(xtx_inv, X.T), y)

#7.8 - What is the sum of all elements of the result?
sum(w)

#Answer = 0.369
```
