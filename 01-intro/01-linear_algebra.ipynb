```python
#We will now create a function that does vector-vector multiplication
def vector_vector_multiplication(u, v):
    """
    This function takes two vectors u and v as input and returns their dot product.
    
    Parameters:
    u (numpy array): First vector
    v (numpy array): Second vector
    
    Returns:
    float: Dot product of u and v
    """
    assert u.shape[0] == v.shape[0] #Ensure both vectors have same length

    n = u.shape[0]
    result = 0.0
    for i in range(n):
        result = result + u[i] * v[i]
    return result

#Run the function
z = vector_vector_multiplication(u, v)

#Or use numpy's built-in function
z_np = np.dot(u, v)

print(z)
print(z_np)
```

```python
#Let's build a function for matrix-vector multiplicaiton

def matrix_vector_multiplication(U, v):
    """
    This function takes a matrix U and a vector v as input and returns their product.
    
    Parameters:
    U (numpy array): Matrix
    v (numpy array): Vector
    
    Returns:
    numpy array: Product of U and v
    """
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]

    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)
    return result

a = matrix_vector_multiplication(U,v)    

print(a)
   
```

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0]
])

def matrix_matrix_multiplication(U, V):
    """
    This function takes a matrix U and a vector v as input and returns their product.
        
        Parameters:
        U (numpy array): Matrix
        V (numpy array): Matrix
        
        Returns:
        numpy array: Product of U and v
    """    
    assert U.shape[1] == V.shape[0]

    num_columns = V.shape[1]
    num_rows = U.shape[0]

    result = np.zeros((num_rows, num_columns))

    for i in range(num_columns):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi

    return result

print(U)
print(V)
a = matrix_matrix_multiplication(U, V)

print(a)

b = np.dot(U,V)

print("b is: ", b)
```
